<h4>To implement data pre-processing and feature engineering techniques on a dataset including removing unnecessary features, feature
scaling and encoding categorical variables using Python libraries.</h4>

<h2>1: Import the basic libraries</h2>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

<h2>2: Load & Analyze the Dataset</h2>

```python
thrilokya_128 = pd.read_csv('diabetes_risk_dataset.csv')
```

```python
thrilokya_128.info()
```

```python
thrilokya_128.head()
```

```python
thrilokya_128.shape
```

```python
# Check for any Null Values
thrilokya_128.isnull().sum()
```

<h2> 3: Remove Unnecessary Features</h2>

```python
# Removing patient_id column, since it does not add predictive value
thrilokya_128 = thrilokya_128.drop('patient_id', axis=1)
```

```python
thrilokya_128.shape
```

<h2> 4: Scale the Numerical Features</h2>

```python
thrilokya_128.info()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
z_scaler = StandardScaler()
```

```python
thrilokya_128[['age', 'blood_pressure', 'insulin_level']] = z_scaler.fit_transform(
    thrilokya_128[['age', 'blood_pressure', 'insulin_level']]
)
```

```python
thrilokya_128.head(3)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
mm_scaler = MinMaxScaler()
```

```python
thrilokya_128[['bmi', 'glucose_level']] = mm_scaler.fit_transform(
    thrilokya_128[['bmi', 'glucose_level']]
)
```

```python
thrilokya_128.head(3)
```

<h2> 5: Encode the Categorical Features</h2>

```python
from sklearn.preprocessing import LabelEncoder
# Create an Instance
encoder = LabelEncoder()
```

```python
thrilokya_128['gender'] = encoder.fit_transform(thrilokya_128['gender'])

thrilokya_128['family_history'] = encoder.fit_transform(thrilokya_128['family_history'])
```

```python
thrilokya_128.head(3)
```

```python
pd.get_dummies(thrilokya_128)
```

<h3>Apply One Hot Encoding using pd.get_dummies()</h3>


```python
pd.get_dummies(thrilokya_128)
```

```python
# Find Boolean columns
bool_col = thrilokya_128.select_dtypes(include='bool').columns

# Convert Boolean columns into integers
for col in bool_col:
    thrilokya_128[col] = thrilokya_128[col].astype(int)

thrilokya_128.info()
```

```python
# Processed Dataset
thrilokya_128.shape
```

```python
# Save Processed Dataset if necessary
thrilokya_128.to_csv('processed_dataset.csv', index=False)
```

<h2>RESULT:</h2>
<h4>The dataset was pre-processed by removing unnecessary features, scaling numerical columns using StandardScaler and MinMaxScaler,
and encoding categorical variables using Label Encoding and One-Hot Encoding. The dataset is now transformed into a clean numerical
form, ready for use in machine learning models.
