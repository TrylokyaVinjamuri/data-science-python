```python
# importing NumPy & Pandas
import numpy as np
import pandas as pd
```

<h2> 2: Create a 2-D array</h2>

```python
# Considering a 4 weeks bike sales data in three different branches 
# rows --> branches, columns --> weeks

sales_thrilokya_128 = np.array([[160, 220, 140, 100],
                 [140, 260, 110, 290],
                [190, 320, 280, 250]])
```

```python
# sales array properties

sales_thrilokya_128.shape, sales_thrilokya_128.ndim
```

<h2> 3: Converting the created 2-D array into DataFrame using Pandas</h2>

```python
df = pd.DataFrame(sales_thrilokya_128) 
df
```

```python
# Creating the Index & Columns for the DataFrame 
# Index --> Branches, Columns --> Weeks
df_index = ['Branch_1', 'Branch_2', 'Branch_3'] 
df_columns = ['Week_1', 'Week_2', 'Week_3', 'Week_4']
```

```python
# Assigning the index & columns to the dataframe
df = pd.DataFrame(data = sales_thrilokya_128, index = df_index, columns = df_columns)
```

```python
df
```

<h2>4:Add Rows and Columns</h2>

```python
# Add Branch_4 (Row)
branch_4 = pd.DataFrame({
    'Week_1': [195],
    'Week_2': [265],
    'Week_3': [175], 
    'Week_4': [290], 
    'Week_5': [245] ,
    'Week_6': [290]
}, index=['Branch_4']) 
df = pd.concat([df, branch_4])
df
```
