<h2> 1: Import the required libraries & dataset</h2>

```python
# Import the libraries 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#load the dataset
df_128= pd.read_csv("experiment_5_alternative_petrol_data.csv")
```

```python
## Info of the Dataset
df_128.info()
```

```python
df_128.head()
```

```python
df_128.value_counts('state')
```

```python
# Convert date column to datetime format
df_128['date'] = pd.to_datetime(df_128['date'])
```

```python
# Maharastra Prices
mah_prices = df_128[df_128['state'] == 'Maharashtra'].sort_values('date')
# Tamilnadu Prices
tam_prices = df_128[df_128['state'] == 'Tamil Nadu'].sort_values('date')
# Telangana State Prices 
tel_prices = df_128[df_128['state'] == 'Telangana'].sort_values('date')
# Karnataka Prices
kar_prices = df_128[df_128['state'] == 'Karnataka'].sort_values('date') 
# Delhi Prices
del_prices = df_128[df_128['state'] == 'Delhi'].sort_values('date')
```

```python
# Reading head of the Prices
mah_prices.head()
```

<h2>2: Visualize Data Using Various Plots</h2>
<h3>Bar Plot</h3>
<h4>Petrol Price in Telangana Over Time</h4>

```python
plt.figure(figsize=(30,8), dpi=250)
plt.plot(tel_prices['date'], tel_prices['rate'], marker='x', color='black')
plt.title('Petrol Price Over Time – Hyderabad') 
plt.xlabel('Date') 
plt.ylabel('Rate')
plt.tight_layout() 
plt.show()
```

<h2>Sub Plots</h2>
<h3>Plotting 2 subplots for Maharashtra & Tamil Nadu Petrol Prices over time</h3>

```python
# Creating scatter plots
fig, axes = plt.subplots(1,2, figsize=(20, 4))

# Maharashtra scatter plot
axes[0].scatter(mah_prices['date'], mah_prices['rate'], color='purple') 
axes[0].set_title('Maharashtra Petrol Prices')
axes[0].set_xlabel('Year') 
axes[0].set_ylabel('Rate')

# Maharashtra scatterplot
axes[1].scatter(tam_prices['date'], tam_prices['rate'], color='green') 
axes[1].set_title('Tamil Nadu Petrol Prices') 
axes[1].set_xlabel('Year')
axes[1].set_ylabel('Rate') 

# Show the plot
plt.show()
```

<h3>3:Comparing 5 States Prices in a single layout</h3>

```python
plt.figure(figsize=(15,6), dpi=200)

plt.plot(mah_prices['date'], mah_prices['rate'], color='red', label='Maharastra')
plt.plot(tam_prices['date'], tam_prices['rate'], color='green', label='Tamilnadu') 
plt.plot(tel_prices['date'], tel_prices['rate'], color='blue', label='Telangana')
plt.plot(kar_prices['date'], kar_prices['rate'], color='yellow', label='Karnataka') 
plt.plot(del_prices['date'], del_prices['rate'], color='orange', label='Delhi') 

plt.title('Petrol Price Over Time')
plt.xlabel('Date')
plt.ylabel('Rate') 
plt.tight_layout() 
plt.legend(loc='upper left', title='States') 
plt.grid(True, linestyle='--') 
plt.show()
```
