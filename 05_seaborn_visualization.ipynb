<h2>MATPLOTLIB</h2>
<h5>>Seaborn is a Statistical library designed to interact with Pandas Dataframes.</h5>
<h5>>It is used to create common statistical plot types.</h5>
<h5>>It uses a simpler one-line syntax -sns.scatterplot(x='x_label',y='y_label',data=dataframe)</h5> 
<h5>>Since, seaborn is built with matplotlib, so still we use plt method to create plots. </h5>
<h5>>Choosing the type of plot depends on the questions and relations we are working with.</h5>
<h5>> While working with seaborn it should be imported - import seaborn as sns</h5>

<h3>1: Import the required libraries & dataset</h3>

```python
# Import the libraries 
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
import numpy as np
```

```python
#load the dataset
df_128 = pd.read_csv("Experiment_6_Alternative_Petrol_Dataset.csv")

```

```python
# Info of the Dataset
df_128.info()
```

```python
df_128.head()
```

```python
# State wise data points in the dataframe 
df_128.value_counts('state')
```

```python
# Convert date column to datetime format
df_128['date'] = pd.to_datetime(df_128['date'])
```

<h2>2: Visualize Data Using Various Plots</h2>
<h4>Scatter Plot</h4>

<h5>Petrol Price in different states Over Time</h5>

```python
plt.figure(figsize=(15,6))
sns.scatterplot(x="date", y="rate", hue="state", data=df_128, alpha=0.8) 
plt.title("Petrol Prices Comparison Across States") 
plt.xticks(rotation=90)
plt.show()
```

<h2>Histograms & KDE</h2>
<h5>Price Distribution with Histograms & KDE</h5>

```python
plt.figure(figsize=(10,4))
sns.kdeplot(data=df_128, x='rate',color='teal')
plt.title("Petrol Price Distribution with KDE");
```

```python
plt.figure(figsize=(10,4))
sns.histplot(data=df_128, x='rate', bins=50, kde=True, color='pink')
plt.title("Petrol Price Distribution with Histogram + KDE");
```

<h2>Count Plot</h2>
<h5> Number of Data Points for each state</h5>

```python
plt.figure(figsize=(6,4)) 
sns.countplot(x="state", data=df_128, hue='state', palette="viridis")
plt.title("Number of Data Points per State") 
plt.xlabel("State")
plt.ylabel("Count") 
plt.show()
```

<h2>Bar plot</h2>

```python
plt.figure(figsize=(6,4))
sns.barplot(x="state", y="rate", data=df_128, hue='state', estimator=np.mean, palette="Set1")
plt.title("Average Petrol Price per State") 
plt.xlabel("State")
plt.ylabel("Average Rate")
plt.show()
```

<h2>Boxplot </h2>
<h5>Finding the IQR's & Outliers of Rate Distributions in different states using Boxplots</h5>

```python
plt.figure(figsize=(6,4), dpi=100)
sns.boxplot(x="rate", y="state", data=df_128, hue='state', palette="Set2")
plt.title("Boxplot of Petrol Prices per State")
plt.xlabel("Rate")
plt.ylabel("State") 
plt.show()
```

<h2>Swarm/ Strip Plots</h2>
<h5>Finding Rate Distributions in different states using Swarm/ Strip Plots</h5>

```python
plt.figure(figsize=(10,4), dpi=150)
sns.stripplot(x="state", y="rate", data=df_128, hue='state', palette="Set1", alpha=0.5)
#sns.swarmplot(x="state", y="rate", data=df, hue='state', palette="Set1", alpha=0.5) 
plt.title("Stripplot of Petrol Prices per State")
plt.xlabel("State")
plt.ylabel("Rate") 
plt.show()
```

<h2>Joint Plots </h2>
<h5> Finding Distributions with Joint Plots</h5>

```python
plt.figure(figsize=(18,8), dpi=150)
sns.jointplot(x="date", y="rate", data=df_128)
plt.suptitle("Jointplot: Rate vs Date") 

plt.xticks(rotation=90) 
plt.show()
```

<h2>Heat Maps </h2>
<h5> Average Price per year per state with Heatmaps</h5>

```python
# Convert date column to datetime
df_128['date'] = pd.to_datetime(df_128['date'])

# Create year column
df_128['year'] = df_128['date'].dt.year
# Average price per year per state
yearly_avg = df_128.groupby(['year','state'])['rate'].mean()
yearly_avg
```

```python
# The 'unstack()' method reshapes the result to make states as columns,
# producing a clean table where each row = year and each column = state
yearly_avg = df_128.groupby(['year','state'])['rate'].mean().unstack()
yearly_avg.head()
```

```python
plt.figure(figsize=(6,4), dpi=100) 
sns.heatmap(yearly_avg, cmap="YlGnBu", fmt=".2f") 
plt.title("Yearly Average Petrol Prices by State")
plt.show()
```
